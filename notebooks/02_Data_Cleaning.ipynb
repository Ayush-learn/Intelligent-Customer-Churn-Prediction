```python
df_clean["TotalCharges"] = df_clean["TotalCharges"].replace(" ", np.nan)
```

```python
df_clean = df_clean.dropna()
```

```python
Observation 1

No duplicate records were found in the dataset.

Observation 2

The TotalCharges column was incorrectly stored as an object datatype because it contained blank spaces instead of numerical values.

Observation 3

A total of 11 blank entries were identified in the TotalCharges column.

Observation 4

These blank entries corresponded to customers with zero tenure, indicating newly joined customers who had not yet accumulated any total charges.

Observation 5

Blank spaces were replaced with NaN to correctly identify missing values.

Observation 6

The TotalCharges column was successfully converted from object to float64, making it suitable for numerical analysis and machine learning.

Observation 7

Since only 11 out of 7043 rows (≈0.16%) contained missing values, removing them had a negligible impact on the dataset.

Observation 8

The cleaned dataset now contains 7032 rows with no missing values and correct data types, making it ready for Exploratory Data Analysis (EDA).
```
